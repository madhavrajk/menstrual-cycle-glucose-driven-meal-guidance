# Synthetic forecast-to-meal demo
This example uses a fictional user, a fixed forecast fixture and a recipe from the
project catalog. It demonstrates the handoff and deterministic restriction checks.
It does not run a trained forecast model or an LLM. Replace the fixture with
Madhav's forecast API when it is available.

```python
from pathlib import Path
import json
from mealguide.guidance.checks import Preferences, check_plan, eligible
from mealguide.guidance.knowledge import load_recipes

profile_path = Path("synthetic_profiles.json")
if not profile_path.exists():
    profile_path = Path("notebooks") / profile_path
profile = json.loads(profile_path.read_text())[0]
prefs = Preferences(profile["diet"], frozenset(profile["allergies"]), profile["max_active_min"])
forecast = {"source": "synthetic fixture", "next_day_symptom": "fatigue", "probability": 0.70}
print("Illustrative forecast:", forecast)
```

```python
recipes = load_recipes()
choices = eligible(recipes, prefs)
assert choices, "No catalog recipes satisfy this profile"
recipe = choices[0]
assert not check_plan([recipe["id"]], recipes, prefs)
print("Example meal plan (one recipe):", recipe["id"])
print("Hands-on cooking time:", recipe["active_time_min"], "minutes")
print("Recipe selected by dietary eligibility only; no symptom benefit is claimed.")
```
